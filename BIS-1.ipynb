```python
import random
import math


cities = {
    'A': (0, 0),
    'B': (2, 3),
    'C': (5, 2),
    'D': (6, 6),
    'E': (1, 5)
}


def distance(city1, city2):
    x1, y1 = cities[city1]
    x2, y2 = cities[city2]

    return math.sqrt((x2 - x1)**2 + (y2 - y1)**2)



def total_distance(route):
    total = 0

    for i in range(len(route)):
        total += distance(
            route[i],
            route[(i + 1) % len(route)]
        )

    return total



def create_population(size):
    population = []

    for _ in range(size):
        route = list(cities.keys())
        random.shuffle(route)
        population.append(route)

    return population



def selection(population):
    population.sort(key=total_distance)
    return population[:2]



def crossover(parent1, parent2):
    child = parent1[:2]

    for city in parent2:
        if city not in child:
            child.append(city)

    return child



def mutation(route):
    i, j = random.sample(range(len(route)), 2)
    route[i], route[j] = route[j], route[i]


random.seed(10)

population = create_population(20)

for generation in range(100):

    parents = selection(population)

    new_population = []

    for _ in range(20):
        child = crossover(parents[0], parents[1])

        if random.random() < 0.2:
            mutation(child)

        new_population.append(child)

    population = new_population

best_route = min(population, key=total_distance)

print("Best Route:", " -> ".join(best_route + [best_route[0]]))
print("Minimum Distance:", round(total_distance(best_route), 2))
```
